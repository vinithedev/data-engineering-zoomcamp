```python
import pyspark
from pyspark.sql import SparkSession
import pandas as pd
```

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName('test') \
    .config("spark.driver.extraJavaOptions", "-Djava.library.path=C:/tools/hadoop/bin") \
    .getOrCreate()
```

```python
df = spark.read.parquet('../data/raw/yellow/2025/11/*')
```

```python
df = df.repartition(4)
```

```python
output_path = '../data/pq/yellow/2025/11/'

df.write.parquet(output_path, mode='overwrite')
```

```python
from pathlib import Path

files = list(Path(output_path).glob('part-*.parquet'))
sizes_mb = [f.stat().st_size / 1024 / 1024 for f in files]

print(f'{len(files)} files')
for f, size in zip(files, sizes_mb):
    print(f'{f.name}: {size:.1f} MB')
print(f'average: {sum(sizes_mb) / len(sizes_mb):.1f} MB')
```
