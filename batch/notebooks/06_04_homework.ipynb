```python
import pyspark
from pyspark.sql import SparkSession
import pandas as pd
```

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName('test') \
    .config("spark.driver.extraJavaOptions", "-Djava.library.path=C:/tools/hadoop/bin") \
    .getOrCreate()
```

```python
# df = spark.read.parquet('../data/raw/yellow/2025/11/*')
df = spark.read.parquet('../data/pq/yellow/2025/11/')
```

```python
# df.schema
# df.take(5)
```

```python
from pyspark.sql import functions as F

df = df.withColumn(
    'trip_duration_hours',
    F.timestamp_diff('SECOND', F.col('tpep_pickup_datetime'), F.col('tpep_dropoff_datetime')) / 3600
)
```

```python
df.select(F.max('trip_duration_hours')).show()
```
