```python
import gzip
import shutil
from pathlib import Path

import pandas as pd

gz_path = Path(r"D:\data-engineering\data-engineering-zoomcamp\batch\data\raw\green\2020\01\green_tripdata_2020_01.parquet.gz")
parquet_path = gz_path.with_suffix("")             # ...green_tripdata_2020_01.parquet
csv_path = parquet_path.with_suffix(".csv")        # ...green_tripdata_2020_01.csv

# 1. Unzip: .parquet.gz -> .parquet
with gzip.open(gz_path, "rb") as src, open(parquet_path, "wb") as dst:
    shutil.copyfileobj(src, dst)

# 2. Convert: .parquet -> .csv
pd.read_parquet(parquet_path).to_csv(csv_path, index=False)
```

```python
base = Path(r"D:\data-engineering\data-engineering-zoomcamp\batch\data\raw\green\2020")

for gz_path in sorted(base.glob("*/*.parquet.gz")):
    parquet_path = gz_path.with_suffix("")
    with gzip.open(gz_path, "rb") as src, open(parquet_path, "wb") as dst:
        shutil.copyfileobj(src, dst)
    pd.read_parquet(parquet_path).to_csv(parquet_path.with_suffix(".csv"), index=False)
    print("converted", gz_path.parent.name)
```

```python
from pathlib import Path

year = 2020

for month in range(1, 13):
    input_dir = Path(f'../data/raw/green/{year}/{month:02d}')
    output_path = f'../data/pq/green/{year}/{month:02d}/'

    csv_files = sorted(input_dir.glob('*.csv'))
    if not csv_files:
        print(f'skipping {year}/{month:02d}: no CSV files in {input_dir}')
        continue

    print(f'processing data for {year}/{month:02d}')

    df_green = spark.read \
        .option("header", "true") \
        .schema(green_schema) \
        .csv([str(f) for f in csv_files])

    df_green \
        .repartition(4) \
        .write.parquet(output_path, mode='overwrite')
```
